> **Dự án NCKH**: *GIPS-DoRA: Generator-Invariant Provenance Subspace Guided Weight-Decomposed Low-Rank Adaptation*
> 
> **Mục tiêu Notebook**:
> 1. **P (AI Polish)**: Hiệu đính đoạn văn tiếng Việt từ tập Human nguyên bản ở 3 cường độ: `P-light` (giữ nguyên cấu trúc câu), `P-medium` (viết lại mượt mà), `P-heavy` (tái cấu trúc toàn diện).
> 2. **G (AI Generation)**: Trích xuất **Semantic Skeleton** (thực thể, luận điểm, số liệu, độ dài) và sinh độc lập hoàn toàn mà KHÔNG truyền văn bản gốc Human (chống rò rỉ bề mặt).
> 3. **Hybrid Sequences**: Ghép nối các chuỗi nhãn hỗn hợp ($H H | P P | G G$) cho mô hình phát hiện chuỗi câu với đầu CRF.
